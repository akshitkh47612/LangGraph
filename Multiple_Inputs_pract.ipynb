```python
import math
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
      values: List[int]
      name: str
      operation: str
      result: str

def op_on_values(state: AgentState) -> AgentState:
      """Either add or multiply the values together"""

      if state['operation'] == '+':
            state['result'] = f"Hi there {state['name']}, the sum of your values is {sum(state['values'])}"
      elif state['operation'] == '*':
            state['result'] = f"Hi there {state['name']}, the product of your values is {math.prod(state['values'])}"
      else:
            state['result'] = f"Hi there {state['name']}, the operation is not supported"

      return state
```
